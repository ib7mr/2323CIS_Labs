Import (Setup)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Activity 1 — Distribution

```python
rng = np.random.default_rng(42)
values = rng.normal(70, 10, 200)

plt.figure(figsize=(6, 4))
plt.hist(values, bins=12, edgecolor='black')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Score Distribution')
plt.show()
```

Activity 2 — Group comparison

```python
dfg = pd.DataFrame({
    'group': ['A']*4 + ['B']*4,
    'score': [75, 81, 79, 84, 69, 73, 76, 71]
})

means = dfg.groupby('group')['score'].mean()

plt.figure(figsize=(6, 4))
means.plot(kind='bar', rot=0, title='Average Score by Group')
plt.ylabel('Average score')
plt.show()
```

Activity 3 — Relationship and annotation

```python
x = np.array([10, 15, 20, 25, 30, 35, 40])
y = np.array([25, 30, 45, 55, 65, 85, 78])

plt.figure(figsize=(6, 4))
plt.scatter(x, y, color='blue', label='Data Points')

max_idx = np.argmax(y)
max_x, max_y = x[max_idx], y[max_idx]

plt.annotate(
    f'Max ({max_x}, {max_y})',
    xy=(max_x, max_y),
    xytext=(max_x - 5, max_y + 5),
    arrowprops=dict(facecolor='red', shrink=0.05)
)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Relationship between X and Y with Max Annotation')
plt.legend()
plt.show()
```

Student Exercise 1

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].boxplot(values)
axes[0].set_title('Score Distribution (Boxplot)')
axes[0].set_ylabel('Score')

means.plot(kind='bar', ax=axes[1], color=['skyblue', 'orange'], rot=0)
axes[1].set_title('Group Mean Comparison')
axes[1].set_ylabel('Average Score')

plt.tight_layout()
plt.show()
```

Student Exercise 2

```python
plt.figure(figsize=(8, 5))
bars = plt.bar(means.index, means.values, color=['#2b5c8f', '#d95f02'], width=0.5)
plt.title('Mean Performance Comparison across Groups', fontsize=12, fontweight='bold')
plt.xlabel('Student Group', fontsize=10)
plt.ylabel('Average Test Score', fontsize=10)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.7)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 2, f'{yval:.1f}', ha='center', va='bottom')

plt.show()
```
